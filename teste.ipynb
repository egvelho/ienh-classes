```python
# Questão 5

# Leitura do código do crachá
cracha = input("Digite o código do crachá")

# Extração via fatiamento (slicing)
# O código interno está entre o índice 4 e 10 (CRA- é do 0 ao 3)
codigo_interno = cracha[4:10]

# O setor ocupa 3 caracteres após o segundo hífen
setor = cracha[11:14]

# O turno é o último caractere da string
codigo_turno = cracha[-1]

# Lógica de interpretação do turno
if codigo_turno == 'M':
    turno_extenso = "Manha"
elif codigo_turno == 'T':
    turno_extenso = "Tarde"
elif codigo_turno == 'N':
    turno_extenso = "Noite"
else:
    turno_extenso = "Turno Invalido"

# Exibição dos resultados
print(codigo_interno)
print(setor)
print(turno_extenso)
```

```python
# Questão 6

# 1. Leitura da linha de dados brutos
linha_bruta = input("Digite a linha de dados do relatório")

# 2. Separação inicial pelo ponto e vírgula
campos_sujos = linha_bruta.split(';')

# 3. Remoção de espaços em branco (trim) de cada campo usando uma lista auxiliar
campos_limpos = []
for campo in campos_sujos:
    campos_limpos.append(campo.strip())

# 4. Construção da nova string com o separador '/'
# O método join é a forma mais eficiente de unir elementos sem deixar um 
# caractere extra no final da sequência.
resultado_final = "/".join(campos_limpos)

# 5. Exibição do relatório formatado
print(resultado_final)
```

```python
# Questão 9

# 1. Leitura da mensagem original
mensagem = input("Digite a mensagem original")

# 2. Definição dos termos ofensivos e suas respectivas máscaras
# A ordem aqui é importante conforme solicitado na tarefa
termos_ofensivos = ['bobo', 'chato', 'idiota', 'feio']
substitutos = ['****', '*****', '******', '****']

# 3. Processamento da mensagem
# Utilizamos o zip() para percorrer as duas listas simultaneamente
for i in range(len(termos_ofensivos)):
    termo = termos_ofensivos[i]
    mascara = substitutos[i]
    
    # Atualizamos a variável 'mensagem' com o texto substituído
    mensagem = mensagem.replace(termo, mascara)

# 4. Exibição da mensagem higienizada
print(mensagem)
```

```python
# Questão 10

# Leitura dos valores de entrada
vendas_brutas = float(input("Digite as vendas brutas"))
custo_produtos = float(input("Digite o custo dos produtos"))
despesas_operacionais = float(input("Digite as despesas operacionais"))

# Cálculos financeiros
receita_total = vendas_brutas - custo_produtos
total_despesas = despesas_operacionais
saldo_final = receita_total - total_despesas

# Lógica para o status do saldo (Positivo, Negativo ou Neutro)
status_saldo = ""
if saldo_final > 0:
    status_saldo = " Positivo"
elif saldo_final < 0:
    status_saldo = " Negativo"

# Montagem do demonstrativo usando f-strings
# Formatação aplicada:
# {campo:<20} -> Alinhado à esquerda com 20 caracteres
# {valor:>15.2f} -> Alinhado à direita com 15 caracteres e 2 casas decimais
print("Demonstrativo de Fechamento de Caixa")
print(f"{'Receita Total':<20}{receita_total:>15.2f}")
print(f"{'Total de Despesas':<20}{total_despesas:>15.2f}")
print(f"{'Saldo Final':<20}{saldo_final:>15.2f}{status_saldo}")
```

```python
# Questão 13

# 1. Leitura da entrada
entrada = input("Digite o endereço")

# 2. Remoção da máscara de rede (caso exista)
# Dividimos no '_mask' e pegamos apenas a primeira parte
if '_mask' in entrada:
    entrada = entrada.split('_mask')[0]

# 3. Padronização dos delimitadores
# Substituímos o sublinhado por hífen para ter um separador único
dados_limpos = entrada.replace('_', '-')

# 4. Separação dos segmentos
segmentos_brutos = dados_limpos.split('-')

# 5. Processamento e Validação dos Octetos
octetos_validados = []

for i in range(4):
    try:
        # Tenta pegar o segmento na posição i
        valor_str = segmentos_brutos[i]
        valor_int = int(valor_str)
        
        # Valida se está no intervalo [0, 255]
        if 0 <= valor_int <= 255:
            octetos_validados.append(valor_int)
        else:
            octetos_validados.append(0)
            
    except (IndexError, ValueError):
        # Se o segmento não existir (IndexError) ou não for número (ValueError)
        octetos_validados.append(0)

# 6. Saída formatada com .format()
# O padrão {:d} garante que o valor seja tratado como inteiro decimal
ip_padronizado = "{:d}.{:d}.{:d}.{:d}".format(
    octetos_validados[0], 
    octetos_validados[1], 
    octetos_validados[2], 
    octetos_validados[3]
)

print(ip_padronizado)
```

```python
# Questão 14

# Leitura do caminho do diretório
caminho = input("Digite o caminho")

# Verifica se o caminho termina com o separador '/', o que indica diretório
if caminho.endswith('/'):
    print("NULO")
else:
    # Localiza a primeira ocorrência do separador /
    primeiro_barra = caminho.find('/')
    
    # Localiza a última ocorrência do separador /
    ultimo_barra = caminho.rfind('/')
    
    # Extrai o nome do arquivo (trecho após a última barra)
    nome_arquivo = caminho[ultimo_barra + 1:]
    
    # Lógica de classificação do arquivo
    # Verificamos se a extensão está entre as permitidas para executáveis
    if nome_arquivo.endswith(('.exe', '.bin', '.sh')):
        classificacao = "Executavel"
    else:
        classificacao = "Arquivo comum"
    
    # Exibição dos resultados
    print(primeiro_barra)
    print(ultimo_barra)
    print(classificacao)
```

```python
# Questão 15

# 1. Leitura do fragmento bruto
fragmento = input("Digite o fragmento")

# Regra 1: Limpeza Global de Espaços
# O strip() sem argumentos remove espaços, tabs (\t) e quebras de linha (\n)
fragmento = fragmento.strip()

# Regra 2: Tratamento de Pontuação de Borda (.,;:!?)
# Removemos a pontuação e, em seguida, limpamos espaços que possam ter sobrado "por baixo"
pontuacao = ".,;:!?"
fragmento = fragmento.strip(pontuacao).strip()

# Regra 3: Higienização de Marcadores de Ruído (^ ~)
ruido = "^~"
fragmento = fragmento.strip(ruido).strip()

# Regra 4: Remoção de Delimitadores de Seção ([ ])
# Aplicada por último e apenas se o fragmento não estiver vazio
delimitadores = "[]"
fragmento = fragmento.strip(delimitadores).strip()

# Exibição do resultado final higienizado
print(fragmento)
```

```python
# Questão 16

# 1. Leitura da mensagem encriptada
entrada = input("Digite a mensagem")

# 2. Descobrir a chave (quantidade de pontos no final)
# Contamos quantos pontos existem no final da string
chave = 0
for char in reversed(entrada):
    if char == '.':
        chave += 1
    else:
        break

# 3. Limpeza da string: remover os pontos e espaços extras nas bordas
mensagem_sem_pontos = entrada.strip(". ")

# 4. Separar a mensagem em palavras
palavras_encriptadas = mensagem_sem_pontos.split()

palavras_decriptadas = []

# 5. Processamento palavra por palavra
for palavra in palavras_encriptadas:
    # Regra: somente palavras com mais de 3 caracteres foram alteradas
    if len(palavra) > 3:
        palavra_original = ""
        for letra in palavra:
            # Converte a letra para seu código ASCII, subtrai a chave e volta para caractere
            codigo_original = ord(letra) - chave
            palavra_original += chr(codigo_original)
        palavras_decriptadas.append(palavra_original)
    else:
        # Palavras com 3 caracteres ou menos permanecem iguais
        palavras_decriptadas.append(palavra)

# 6. Reconstruir a frase e exibir o resultado
resultado = " ".join(palavras_decriptadas)
print(resultado)
```
